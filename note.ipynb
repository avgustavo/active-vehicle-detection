```python
from pathlib import Path
from ultralytics import YOLO
```

```python
project_name = 'pipeline'
classes = [0,1,2]
```

```python
for i in range(5):
    best_model_path = Path(project_name) / f'ciclo_{i}' / "weights" / "best.pt"
    yaml_path = Path(f'runs/{project_name}/config') / f'ciclo_{i}' / "data.yaml"

    model = YOLO(best_model_path)
    metrics = model.val(
        data=yaml_path,
        split='val',
        name=f'ciclo_{i}_val',
        project=project_name,
        classes=classes
    )

    print(f"  > mAP50-95 (val): {metrics.box.map:.4f}")
    print(f"  > mAP50 (val):    {metrics.box.map50:.4f}")
    print(f"  > mAP (val):      {metrics.box.map75:.4f}")

    metrics_csv = metrics.to_csv()
    with open(f'{project_name}/{cycle_name}/res_val.csv', 'w') as f:
        f.write(metrics_csv)
```
